# Traffic Accident Analysis and Accident Hotspot Detection System

This notebook is dedicated to analyzing traffic accident data and identifying accident hotspots. We will utilize spatial data analysis, statistical modeling, and visualization libraries to detect areas with high accident density.

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import folium
from folium.plugins import HeatMap

# Set plotting style
sns.set_theme(style='whitegrid')
plt.rcParams['figure.figsize'] = (10, 6)

print('Libraries successfully imported!')
```

## 1. Dataset Generation / Loading
We will construct a simulated dataset of traffic accidents in a metropolitan area containing latitudes, longitudes, severity levels, weather, and timestamps.

```python
np.random.seed(42)
n_samples = 500

# Center around a hypothetical city center (e.g., latitude 37.7749, longitude -122.4194 for San Francisco)
city_center = [37.7749, -122.4194]

# Simulate Coordinates with cluster hotspots
latitudes = np.concatenate([
    np.random.normal(37.7749, 0.015, int(n_samples * 0.5)),
    np.random.normal(37.7900, 0.008, int(n_samples * 0.3)),
    np.random.normal(37.7550, 0.010, int(n_samples * 0.2))
])

longitudes = np.concatenate([
    np.random.normal(-122.4194, 0.015, int(n_samples * 0.5)),
    np.random.normal(-122.4000, 0.008, int(n_samples * 0.3)),
    np.random.normal(-122.4400, 0.010, int(n_samples * 0.2))
])

data = pd.DataFrame({
    'Latitude': latitudes,
    'Longitude': longitudes,
    'Severity': np.random.choice(['Low', 'Medium', 'High'], size=n_samples, p=[0.5, 0.35, 0.15]),
    'Weather': np.random.choice(['Clear', 'Rainy', 'Foggy', 'Cloudy'], size=n_samples, p=[0.6, 0.2, 0.1, 0.1]),
    'Road_Condition': np.random.choice(['Dry', 'Wet', 'Icy'], size=n_samples, p=[0.7, 0.25, 0.05]),
    'Hour': np.random.randint(0, 24, size=n_samples)
})

print('Dataset generated successfully!')
print(data.head())
```

## 2. Exploratory Data Analysis (EDA)
We will visualize how accidents are distributed across different times of the day, weather, and road conditions.

```python
fig, axes = plt.subplots(2, 2, figsize=(16, 12))

# Accident Severity Distribution
sns.countplot(data=data, x='Severity', hue='Severity', order=['Low', 'Medium', 'High'], palette='viridis', legend=False, ax=axes[0,0])
axes[0,0].set_title('Accident Severity Distribution')

# Hourly Distribution
sns.histplot(data=data, x='Hour', bins=24, kde=True, color='crimson', ax=axes[0,1])
axes[0,1].set_title('Accident Distribution by Hour of Day')
axes[0,1].set_xticks(range(0, 24, 2))

# Weather Conditions
sns.countplot(data=data, x='Weather', hue='Weather', palette='magma', legend=False, ax=axes[1,0])
axes[1,0].set_title('Accidents by Weather Condition')

# Road Conditions
sns.countplot(data=data, x='Road_Condition', hue='Road_Condition', palette='coolwarm', legend=False, ax=axes[1,1])
axes[1,1].set_title('Accidents by Road Condition')

plt.tight_layout()
plt.show()
```

## 3. Geospatial Hotspot Detection
We will map the accidents and layer an interactive heatmap to locate dense clusters (accident hotspots) using `folium`.

```python
# Create interactive folium map centered on the city
m = folium.Map(location=city_center, zoom_start=13, control_scale=True)

# Prepare heatmap data format: list of [lat, lon]
heat_data = data[['Latitude', 'Longitude']].values.tolist()

# Add HeatMap layer
HeatMap(heat_data, radius=15, blur=10, max_zoom=13).add_to(m)

# Display map
m
```
